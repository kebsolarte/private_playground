```python
import random

num_items = 10

scores = []
for item in range(0, num_items):
    random_num = random.randint(1, 100)
    scores.append(random_num)

total = 0 
for score in scores:
    total += score

max = 0
for score in scores:
    if score > max:
        max = score
    
min = 0
for score in scores:
    if min == 0:
        min = score
    elif score < min:
        min = score

print(scores)
print(total)
print(max)
print(min)
```

```python
random.shuffle(scores)
print(scores)
```

```python
class User:
    
    def __init__(self, name, user_id):
        self.name = name
        self.id = user_id
        self.followers = 0
        self.following = 0

    def follow(self, user):
        self.following += 1
        user.followers += 1

user_1 = User("Kevin", "001")
user_2 = User("Shaz", "002")
user_3 = User("Chewy", "003")

print(user_1.name)
print(user_1.id)

user_1.follow(user_2)
user_1.follow(user_3)
user_2.follow(user_3)

print(user_1.following)
print(user_1.followers)
print(user_2.following)
print(user_2.followers)
print(user_3.following)
print(user_3.followers)
```

```python
average = lambda x: sum(x) / len(x)

numbers = [1,2,3,4,5,6,7]

print(average(numbers))


capitalize = lambda x: x.capitalize()

words = ["shaz", "chewy", "kevin"]

capitalized_words = map(capitalize, words)
capitalized_words = list(capitalized_words)

print(capitalized_words)
```

```python
# Iterable
names = ['chewy', 'shaz', 'kevs']

# Creating an interator using iter() method
names_iterator = iter(names)

print(next(names_iterator))
print(next(names_iterator))
```

```python
# Using enumerate to create a list of tuples of index and values
print(list(enumerate(names)))

for index, value in enumerate(names):
    print(index, value)
```

```python
core = ['dog', 'human', 'human']
gender = ['female', 'female', 'male']

zipped = zip(names, core, gender)

print(zipped)
print(list(zipped))
print(zipped)

# Printing consumes elements in the zip object
print(*zipped)

# Recreating the zip object
zipped = zip(names, core, gender)

# Unpacking the zip object and 'unzipping' it
names2, core2, gender2 = zip(*zipped)

print(names2, core2, gender2)
```
