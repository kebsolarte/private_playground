```python
import random

num_items = 10

scores = []
for item in range(0, num_items):
    random_num = random.randint(1, 100)
    scores.append(random_num)

total = 0 
for score in scores:
    total += score

max = 0
for score in scores:
    if score > max:
        max = score
    
min = 0
for score in scores:
    if min == 0:
        min = score
    elif score < min:
        min = score

print(scores)
print(total)
print(max)
print(min)
```

```python
random.shuffle(scores)
print(scores)
```
