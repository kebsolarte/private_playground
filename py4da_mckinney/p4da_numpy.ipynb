```python
# Numpy enabls batch computations of arrays with similar syntax as scalar objects
#  ndarray is a generic multidimensional container for homogenous data
import numpy as np

data = np.array([range(0, 5), range(5, 10)])
print(data)

# np.arange() is similar to range function
data2 = np.arange(5)
print(data2)

# Operations between equal-size arrays apply the operation element-wise
print(data * 10)
print(data + data)
print(data - data)
print(data == data)

# shape - tuple indicating the size of each dimension
print(data.shape)

# dtype - object describing the data type of the array
print(data.dtype)
```

```python
# Specifying dtype
arr1 = np.array([1, 2, 3], dtype=np.float64)
arr2 = np.array([1, 2, 3], dtype=np.int32)

print(arr1)
print(arr2)

# Converting dtypes
arr1 = arr1 + 0.5
print(arr1)
arr1 = arr1.astype(np.int32)
print(arr1)
```

```python
# Slicing -> in contrast with lists, ndarray slices are views of arrays and do not create copies
arr1_slice = arr1[1:4]
print(arr1_slice)
arr1_slice[1] = 5
print(arr1)

# Slicingg 2D arrays
data_slice = data[1]
print(data_slice)
print(data[1,2])
print(data[1:, 2:])
print(data[:1, 2:])
print(data[:2, 2])
```
