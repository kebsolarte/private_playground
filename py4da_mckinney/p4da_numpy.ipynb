```python
# Numpy enabls batch computations of arrays with similar syntax as scalar objects
#  ndarray is a generic multidimensional container for homogenous data
import numpy as np

data = np.array([range(0, 5), range(5, 10)])
print(data)

print(data * 10)

print(data + data)

# shape - tuple indicating the size of each dimension
print(data.shape)

# dtype - object describing the data type of the array
print(data.dtype)
```
