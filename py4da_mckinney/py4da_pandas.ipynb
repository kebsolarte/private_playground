```python
import pandas as pd
import numpy as np
```

```python
# Series
ser1 = pd.Series([4, -3, -6, 6, 7])
print(ser1)
print(ser1.array)
print(ser1.index)

# You can assign the index
ser2 = pd.Series([2, 6, -7, 3], index=['d', 'b', 'c', 'a'])
print(ser2)
print(ser2.index)

# Calling a data point
print(ser2['a'])
print(ser2['d'])
print('a' in ser2)

ser2['c'] = 4
print(ser2['c'])
```

```python
# Creating a series from a dict
data = {'Baguio': 126357, 'Bulacan': 7863826, 'Pampanga': 347628176, 'Manila': 7182318}

data_ser = pd.Series(data)
print(data_ser)

# Convert back
data_dict = data_ser.to_dict()
print(data_dict)

# Enforcing keys as indices. Indices with no values are assigned NaN, indices not included are dropped
cities = ['Pampanga', 'Baguio', 'Bulacan', 'Pasig']
data_ser2 = pd.Series(data, index=cities)
print(data_ser2)

pd.isna(data_ser2)
pd.notna(data_ser2)

# Setting names of Series and indices
data_ser2.name = 'population'
data_ser2.index.name = 'cities'

print(data_ser2)
print(data_ser2.index)
```

```python
# DataFrame
data ={
    'city': ['Baguio', 'Bulacan', 'Pampanga', 'Baguio', 'Bulacan', 'Pampanga'],
    'year': ['2000', '2000', '2000', '2001', '2001', '2001'],
    'pop': [1.2, 1.4, 1.6, 2.3, 2.7, 2.4]
}

frame = pd.DataFrame(data)
print(frame)
print(frame.head())

# Rearranging the columns, columns not in the dict will have NaN values
frame2 = pd.DataFrame(data, columns=['year', 'city', 'pop', 'debt'])
print(frame2)

# Columns can be modified by assignment
frame2['debt'] = np.arange(100_000, 700_000, 100_000)
frame2['gdp'] = 100_000_000
print(frame2)
```

```python
# In nested dicts, inner keys are interpreted as indices and outer keys as columns
populations = {
    'Baguio': {2000: 1.3, 2001: 1.6, 2002: 3.5},
    'Bulacan': {2000: 1.6, 2001: 1.9, 2002: 2.5},
    'Pampanga': {2000: 1.8, 2001: 1.0, 2002: 4.5, 2004: 5.2}
}

frame3 = pd.DataFrame(populations)
print(frame3)

# Transposing is similar to numpy
# Note that transposing may discard column data types which are lost when you transpose it back
frame4 = frame3.T 
print(frame4)
```

```python
# Index objects
# Rows and columns are treated as index objects, which are objects responsible for holding axis labels and metadata
# Index objects behave as set objects
frame3.index.name = 'year'
frame3.columns.name = 'cities'

print(frame3.columns)
print(frame3.index)

print('Baguio' in frame3.columns)
```

```python
# Arithmetic operations
df1 = pd.DataFrame(
    np.arange(12).reshape((3,4)),
    columns=list('abcd')
    )

df2 = pd.DataFrame(
    np.arange(20).reshape((4,5)),
    columns=list('abcde')
    )

# Placing a NaN inside
df2.loc[1, 'b'] = np.nan

print(df1)
print(df2)
```

```python
# Adding DFs will results to a union of the DFs with NaN as values for non-overlapping elements
total = df1 + df2
print(total)

# You can use the fill_value argument of add method to fill in missing values
total1 = df1.add(df2, fill_value=0)
print(total1)

# You can also try to reindex one table first before peforming the operation to align the location of the elements
reindexed_df1 = df1.reindex(columns=df2.columns, index=df2.index, fill_value=0)
print(reindexed_df1)
```

```python
# Operations between DFs and series
ser1 = np.arange(4)

# The series elements are subtracted once for each row, this is referred to as broadcasting
diff = df1 - ser1
print(diff)
```

```python
# Broadcasting over the columns, matching on the rows, will need to specify the index
df3 = df1.reindex(index=df2.index, fill_value=0)

diff2 = df3.sub(ser1, axis='index')
print(diff2)
```

```python
# Function application and mapping
frame = pd.DataFrame(
    np.random.standard_normal((4,3)),
    columns=list('bed'),
    index=['Baguio', 'Bulacan', 'Pampanga', 'Manila']
)

print(frame)

# NP ufuncs also work with pandas objects
frame2 = np.abs(frame)
print(frame2)
```

```python
# Functions on 1D arrays can also be applied to rows or columns using DF apply method
# Functions for axis-wise execution, use apply
def func(x):
    return x.max() - x.min()

frame.apply(func)
```

```python
# To apply func across columns, use axis argument
frame.apply(func, axis='columns')
```

```python
# Functions applied through apply can also return a series with multiple values
def func2(x):
    return pd.Series([x.min(), x.max(), func(x)], index=['min', 'max', 'diff'])

frame.apply(func2)
```

```python
# Stat functions like sum and mean are DF methods and can be pulled without using apply
print(frame.sum())
print(frame.mean())
```

```python
# Functions for element-wise execution, use applymap
def format(x):
    return f'{x:.2f}'

frame.applymap(format)

# Modern implementation use map method, which yields the same output
frame.map(format)
```

```python
# To sort by row or column, use the sort_index(axis=) method which returns a new object
index_sorted = frame.sort_index()
index_sorted
```

```python
column_sorted = frame.sort_index(axis='columns', ascending=False)
column_sorted
```

```python
# To sort by values, use the sort_values method instead
reindex_frame = frame.reindex(index=['Baguio', 'Bulacan', 'Pampanga', 'Manila', 'Baler'])
print(reindex_frame)

sorted_reindex_frame = reindex_frame.sort_values('b')
print(sorted_reindex_frame)

sorted_reindex_frame = reindex_frame.sort_values(['b', 'd'], na_position='first')
print(sorted_reindex_frame)
```

```python
# Ranking assigns rank and breaks ties by assigning the mean rank
ser1 = pd.Series([5, 3, 0, 2, 4, 4, -8, -7, 6])

ser1.rank()
```

```python
print(ser1.sort_values())
ser1.sort_values().rank(ascending=False)
```

```python
# DFs can compute ranks over either rows or columns
print(frame.rank(axis='columns'))
print(frame.rank(axis='index'))
```

```python
# Descriptive stats
df = pd.DataFrame(
    [[1.4, np.nan], [7.1, -4.5],
     [np.nan, np.nan], [0.75, -1.31]],
     index=list('abcd'),
     columns=['one', 'two']
)
df
```

```python
# Built in methods can handle NaN values
print(df.sum())
print(df.sum(axis='columns'))
print(df.sum(axis='columns', skipna=False))
```

```python
# Some aggs like mean require at least one non-NA value
df.mean(axis='columns')
```

```python
# Some methods return indirect statistics like the index value
print(df.idxmax())

# Others are accumulations
print(df.cumsum())

# You can use describe method to produce summary statistics in one shot
print(df.describe())

# You can also use describe for non-numeric data
obj = pd.Series(list('abcd')*4)
obj.describe()
```

```python
# Getting unique values in a series
print(obj.unique())

# Getting value frequencies
print(obj.value_counts())

# Checking set membership, which is useful in filtering a dataset
print(obj.isin(['b', 'c']))
print(obj[obj.isin(['b', 'c'])])
```
