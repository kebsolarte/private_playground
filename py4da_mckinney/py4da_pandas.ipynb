Notebook for learning Pandas

```python
import pandas as pd
import numpy as np
```

```python
# Series
ser1 = pd.Series([4, -3, -6, 6, 7])
print(ser1)
print(ser1.array)
print(ser1.index)

# You can assign the index
ser2 = pd.Series([2, 6, -7, 3], index=['d', 'b', 'c', 'a'])
print(ser2)
print(ser2.index)

# Calling a data point
print(ser2['a'])
print(ser2['d'])
print('a' in ser2)

ser2['c'] = 4
print(ser2['c'])
```

```python
# Creating a series from a dict
data = {'Baguio': 126357, 'Bulacan': 7863826, 'Pampanga': 347628176, 'Manila': 7182318}

data_ser = pd.Series(data)
print(data_ser)

# Convert back
data_dict = data_ser.to_dict()
print(data_dict)

# Enforcing keys as indices. Indices with no values are assigned NaN, indices not included are dropped
cities = ['Pampanga', 'Baguio', 'Bulacan', 'Pasig']
data_ser2 = pd.Series(data, index=cities)
print(data_ser2)

pd.isna(data_ser2)
pd.notna(data_ser2)

# Setting names of Series and indices
data_ser2.name = 'population'
data_ser2.index.name = 'cities'

print(data_ser2)
print(data_ser2.index)
```

```python
# DataFrame
data ={
    'city': ['Baguio', 'Bulacan', 'Pampanga', 'Baguio', 'Bulacan', 'Pampanga'],
    'year': ['2000', '2000', '2000', '2001', '2001', '2001'],
    'pop': [1.2, 1.4, 1.6, 2.3, 2.7, 2.4]
}

frame = pd.DataFrame(data)
print(frame)
print(frame.head())

# Rearranging the columns, columns not in the dict will have NaN values
frame2 = pd.DataFrame(data, columns=['year', 'city', 'pop', 'debt'])
print(frame2)

# Columns can be modified by assignment
frame2['debt'] = np.arange(100_000, 700_000, 100_000)
frame2['gdp'] = 100_000_000
print(frame2)
```

```python
# In nested dicts, inner keys are interpreted as indices and outer keys as columns
populations = {
    'Baguio': {2000: 1.3, 2001: 1.6, 2002: 3.5},
    'Bulacan': {2000: 1.6, 2001: 1.9, 2002: 2.5},
    'Pampanga': {2000: 1.8, 2001: 1.0, 2002: 4.5, 2004: 5.2}
}

frame3 = pd.DataFrame(populations)
print(frame3)

# Transposing is similar to numpy
# Note that transposing may discard column data types which are lost when you transpose it back
frame4 = frame3.T 
print(frame4)
```

```python
# Index objects
# Rows and columns are treated as index objects, which are objects responsible for holding axis labels and metadata
# Index objects behave as set objects
frame3.index.name = 'year'
frame3.columns.name = 'cities'

print(frame3.columns)
print(frame3.index)

print('Baguio' in frame3.columns)
```
