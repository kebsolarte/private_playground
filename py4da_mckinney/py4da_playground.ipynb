```python
# Review of some python basics
# Use of zip to pair sequences. It returns the shortest sequence
seq1 = ['Kev', 'Shaz', 'Chewy']
seq2 = ['Human', 'Human', 'Dog', 'Cat']

zipped = zip(seq1, seq2)
print(zipped)           # Just prints pointer
print(next(zipped))     # Zipped object is an iterator object, returning and consuming pairs one at a time
print(list(zipped))     # Once used, the iterator is exhausted

# Creating dictionaries via zip
items = {key: value for key, value in zip(seq1, seq2)}
print(items)
```

```python
# Nested comprehensions
list_of_colors = [['green', 'blue', 'indigo', 'violet'], ['red', 'orange', 'yellow']]

# Retrieving 6-letterd colors
# Instead of writing it this way
sample1 = []
for colors in list_of_colors:
    for color in colors:
        if len(color) == 6:
            sample1.append(color)
print(sample1)

# You can use a simple list comprehension to be more concise
sample2 = []
for colors in list_of_colors:
    colors = [color for color in colors if len(color) == 6]
    sample2.extend(colors)
print(sample2)

# Or you can write a nested list comprehension. Yikes.
sample3 = [color for colors in list_of_colors for color in colors if len(color) == 6]
print(sample3)
```

```python
# Playing with functions
raw_provinces = [' Baguio', 'bUlacAn', 'pampanga! ', ' bataAn? ']

# Cleaning data
import re 

def clean_data(raw_data):
    cleaned_data = []
    for item in raw_data:
        item = item.strip()
        item = item.title()
        item = re.sub('[!?]', '', item)
        cleaned_data.append(item)
    return cleaned_data

cleaned_data1 = clean_data(raw_provinces)
print(cleaned_data1)

# Another way is to make a list of operations to apply
def remove_punctuations(word):
    return re.sub('[!?]', '', word)

operations = (str.strip, str.title, remove_punctuations)

cleaned_data2 = []
for province in raw_provinces:
    for func in operations:
        province = func(province)
    cleaned_data2.append(province)
print(cleaned_data2)

# Or use map function
def cleaning_series(string):
    for func in operations:
        string = func(string)
    return string
 
cleaned_data3 = [province for province in map(cleaning_series, raw_provinces)]
print(cleaned_data3)

# Alternatively, we can write a list comprehension
cleaned_data4 = [cleaning_series(province) for province in raw_provinces]
print(cleaned_data4)
```

```python
# Working with lambda functions
ints = [1, 2, 3, 4, 5]

def apply_to_list(list, func):
    return [func(item) for item in list]

sample1 = apply_to_list(ints, lambda x: x * 2)
print(sample1)

# You can also pass a lambda expression in a method
random = ['adfaf', 'ruytenxbcv', 'aaaa', 'ahsgajhsg', 'abababab', 'werew']

# Sorting based on num of unique letters
random.sort(key=lambda x: len(set(x)))
print(random)
```

```python
# Working with generators
# A generator is a way to construct an iterable object which returns items in a lazy execution = less memory
def doubles(n=10):
    print(f"Generating the doubles of 1 to {n}...")
    for n in range(n):
        yield n * 2     # Yield keyword is used instead of return

# Nothing happens the first time a generator is called
gen = doubles()
print(gen)

# Elements are returned once the generator executes its code
for x in gen:
    print(x, end=' ')

print()

# Another way is to create generator expression, which are somewhat equivalents of list comprehensions, but uses () instead of []
gen = (x * 2 for x in range(10))
print(gen)

# These expressions can be used as function arguments in some cases
print(sum(gen))

# But, as an iterable, elements are exhausted after execution, so you have to materialize the generator object again
gen = (x * 2 for x in range(10))
print({(x/2, x) for x in gen})
```
