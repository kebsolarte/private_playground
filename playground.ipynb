```python
class User:
    
    def __init__(self, name, user_id):
        self.name = name
        self.id = user_id
        self.followers = 0
        self.following = 0

    def follow(self, user):
        self.following += 1
        user.followers += 1

user_1 = User("Kevin", "001")
user_2 = User("Shaz", "002")
user_3 = User("Chewy", "003")

print(user_1.name)
print(user_1.id)

user_1.follow(user_2)
user_1.follow(user_3)
user_2.follow(user_3)

print(user_1.following)
print(user_1.followers)
print(user_2.following)
print(user_2.followers)
print(user_3.following)
print(user_3.followers)
```

```python
# Generators - does not store objects or items in memory unlike list comprehensions, but is an iterable
# Similar to list comprehension but created using ()

doubles = (i * 2 for i in range(10))
print(doubles)

# Lazy evaluation - useful for generating sequences on the fly, since generators does not store items in memory
print(next(doubles))
print(next(doubles))
print(next(doubles))

# Example 2
# Create generator object: result
result = (i for i in range(0,31))

# Print the first 5 values
print('----------------------')
print(next(result))
print(next(result))
print(next(result))
print(next(result))
print(next(result))
print('----------------------')

# Print the rest of the values
for value in result:
    print(value)

```

```python
# Generator functions - similar to a normie fx but uses yield instead of return
# Create a list of strings
lannister = ['cersei', 'jaime', 'tywin', 'tyrion', 'joffrey']

# Define generator function get_lengths
def get_lengths(input_list):
    """Generator function that yields the
    length of the strings in input_list."""

    # Yield the length of a string
    for person in input_list:
        yield len(person)

generator = get_lengths(lannister)

print(next(generator))
print(next(generator))
print('---------')

# Print the values generated by get_lengths()
for value in get_lengths(lannister):
    print(value)
```

```python
# Working with sets

cookie_tray_1 = ['chocolate chip', 'white chocolate', 'biscotti', 'dark chocolate', 'biscotti', 'dark chocolate']
cookie_tray_2 = ['chocolate chip', 'anzac', 'digestive', 'white chocolate', 'chocolate chip']

cookie_set_1 = set(cookie_tray_1)
cookie_set_2 = set(cookie_tray_2)

print(cookie_set_1)
print(cookie_set_2)

# Usual set methods depend on the sequence and placement of sets
print(cookie_set_1.difference(cookie_set_2))
print(cookie_set_2.difference(cookie_set_1))

# Set theory
print(cookie_set_1.union(cookie_set_2))
print(cookie_set_1.intersection(cookie_set_2))

# Modifying sets
cookie_set_1.add('white chocolate')
print(cookie_set_1)

cookie_set_1.add('matcha')
print(cookie_set_1)

cookie_set_2.update(['chocolate chip', 'dark chocolate', 'matcha'])
print(cookie_set_2)

# Removing elements
cookie_set_1.discard('biscotti')
print(cookie_set_1)

# Remove and return arbitrary element
print(cookie_set_1.pop())
print(cookie_set_1)
```

```python
from palmerpenguins import load_penguins
import pandas as pd
from collections import Counter

penguins = load_penguins()
penguins.head()

# Using counter data objects which returns a dict of elements and counts
penguins_species_count = Counter(penguins['species'])
print(penguins_species_count)

# using the most_common method
print(penguins_species_count.most_common(2))
```

```python
# Using defaultdict, which initializes a dict with value type defined
from collections import defaultdict

penguins_subset = penguins[['species', 'island', 'body_mass_g', 'sex', 'flipper_length_mm']]

penguins_by_island = defaultdict(list)

# Iterate over each row
for index, row in penguins_subset.iterrows():
   penguins_by_island[row['island']].append((row['species'], row['sex'], row['body_mass_g'], row['flipper_length_mm']))

print(penguins_by_island)
```

```python
from collections import namedtuple

# namedtuples allow access to elements uisng the attributes similar to dictionaries
TorgersenPenguins = namedtuple('TorgersenPenguins', ['species', 'sex', 'mass'])

# Create a list of Torgersen penguin details as tuples 
torgersen_penguins = []
for index, row in penguins_subset.iterrows():
    if row['island'] == 'Torgersen':
        details = TorgersenPenguins(row['species'], row['sex'], row['body_mass_g'])
        torgersen_penguins.append(details)

# Using namedtuples
for penguin in torgersen_penguins:
    if penguin.species == 'Adelie':
        print(penguin.sex, penguin.mass)
```

```python
# Using dataclasses to store more complex prperties
from dataclasses import dataclass

# Defining a dataclass with straightforward properties using the dataclass decorator
@dataclass
class Penguins:
    species: str
    island: str
    sex: str
    mass: int
    flipper_length: int

# The property decorator can be used for more cplex custom properties
    @property
    def flipper_to_mass_ratio(self):
        return self.flipper_length / self.mass

penguins_class = []
for index, row in penguins_subset.iterrows():
    penguins_class.append(Penguins(row['species'], row['island'], row['sex'], row['body_mass_g'], row['flipper_length_mm']))

print(penguins_class[0:5])

# Accessing dataclass properties
for penguin in penguins_class[0:2]:
    print(penguin.species)
    print(penguin.island)
    print(penguin.sex)
    print(penguin.mass)
    print(penguin.flipper_length)
    print(penguin.flipper_to_mass_ratio)
    print()

    
```
