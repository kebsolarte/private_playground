```python
names = ['Alex', 'Beth', 'Caroline', 'Dave', 'Eleanor', 'Freddie']

short_names = [name for name in names if len(name) <= 4]
print(short_names)

capitalized_long_names = [name.upper() for name in names if len(name) > 4]
print(capitalized_long_names)
```
